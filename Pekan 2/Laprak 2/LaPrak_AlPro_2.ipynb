## 2.1 Dasar Teori
1. Konsep dasar variabel Python

Variabel adalah nama yang digunakan untuk merujuk pada suatu nilai atau objek yang disimpan di memori. Variabel memudahkan programmer dalam mengakses dan mengolah data tanpa harus menulis nilai secara langsung berulang kali, dan di Python kita hanya cukup memberi nilai dengan tanda = tanpa mendeklarasikan tipe datanya. Ini karena Python memiliki sistem *dynamic typing* yang mana tipe data variabel akan ditentukan secara otomatis oleh Python berdasarkan nilai yang diberikan.

2. Tipe data dasar

String (str) = Teks yang ditulis dengan tanda kutip. Contoh = "Python"

Integer (int) = Bilangan bulat tanpa desimal. Contoh = 1337

Float (float) = Bilangan desimal. Contoh = 3.17

Boolean (bool) = Tipe data yang menyimpan nilai True atau False. 

3. Konstanta

Tempat atau variabel dengan nilai tetap. Di Python, penulisan nama konstanta biasanya menggunakan huruf kapital semua (contoh: PI = 3.14) Digunakan untuk menyimpan data yang tidak boleh dirubah.


## 2.2 Kode Program

### 2.2.1 Program "String_Nim"

```python
nama_2008 = "Rafi"
jenis_kelamin_2008 = "L"
alamat_2008 = """   
   Gunung Pangilun
   Kec.Padang Utara
   Kota Padang
"""
agama_2008 = "Islam"
print ("Nama saya adalah: ", nama_2008)
print ("jenis kelamin saya adalah: ", jenis_kelamin_2008)
print ("Alamat saya adalah: ", alamat_2008)
print ("agama saya adalah: ", agama_2008)
```


**Penjelasan :**
1. Barisan pertama sampai kedua berfungsi untuk menyimpan data nama dan kelamin
2. Barisan ketiga sampai ketujuh berfungsi untuk menyimpan data alamat yang terdiri dari 3 baris.
3. Barisan delapan berfungsi untuk menyimpan data agama.
4. Barisan sembilan sampai dua belas berfungsi untuk menampilkan nama, jenis kelamin, alamat dan agama ke layar.


### 2.2.2 Program "Numerik_Nim"

```python
a_2008=100
print ("The type of variable having value", a_2008, " is ", type(a_2008))

b_2008=10.2345
print ("The type of variable having value", b_2008, " is ", type (b_2008))

c_2008=100+3j
print ("The type of variable having value", c_2008, " is ", type (c_2008))
```


**Penjelasan :**
1. Barisan pertama sampai kedua berfungsi untuk membuat variabel a_2008 dengan nilai 100 dan juga menampilkan tipe datanya dengan fungsi type ().
2. Barisan empat sampai lima berfungsi untuk membuat variabel b_2008 dengan nilai 10.2345 serta menampilkan tipe datanya.
3. Barisan tujuh sampai delapan berfungsi untuk membuat variabel c_2008 dengan nilai 100+3j serta menampilkan tipe datanya.

### 2.2.3 Program "String_2611532008"

```python
is_lulus_2008 = True
is_cumlaude_2008 = True

nilai_2008 = 80
batas_lulus_2008 = 75

status_kelulusan_2008 = nilai_2008 >= batas_lulus_2008  # Hasilnya akan true

print ("=== Check Kelulusan ===")
print ("Nilai:", nilai_2008)
print("Apakah lulus?:", status_kelulusan_2008)
if is_lulus_2008 and is_cumlaude_2008:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```


**Penjelasan :**
1. Barisan pertama sampai kedua berfungsi untuk menyimpan variabel is_lulus_2008 dengan nilai true dan is_cumlaude_2008 bernilai true.
2. Barisan empat sampai lima berfungsi untuk menyimpan nilai mahasiswa dan batas nilai kelulusan.
3. Barisan tujuh berfungsi untuk memeriksa kalau nilai mahasiswa lebih tinggi atau sama dengan batas nilai kelulusan, jika hasilnya sama atau tinggi maka hasilnya akan true.
4. Barisan sembilan sampai tiga belas berfungsi untuk menampilkan judul, nilai, status kelulusan, dan jika mahasiswa cum laude ke layar. 

### 2.2.3 Program "Input_2611532008"

```python
panjang_2008 = input('Masukan nilai panjang: ')
lebar_2008 = input('Masukan nilai lebar: ')
luas_2008 = int(panjang_2008) * int(lebar_2008)
print ("Luas =", luas_2008)
```


**Penjelasan :**
1. Barisan pertama sampai kedua berfungsi sebagai tempat input nilai panjang dan lebar persegi panjang.
2. Barisan ketiga berfungsi untuk mengubah nilai panjang dan lebar ke nilai integer, dan mengkalikan nilai tersebut sehingga menghasilkan nilai luas persegi panjang.
3. Barisan keempat berfungsi untuk menampilkan nilai luas persegi panjang ke layar.


### 2.2.5 Program "String_2611532008"

```python
from typing import Final

PI_2008: Final[float] = 3.14
print("PI_2008: %f" % (PI_2008))
jari_2008 = float(input("Masukkan jari-jari: "))
luas_2008 = PI_2008 * jari_2008 * jari_2008
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_2008, luas_2008))
```


**Penjelasan :**
1. Barisan pertama sampai ketiga berfungsi untuk import dan menetapkan nilai konstanta PI sebesar 3.14.
2. Barisan keempat sampai kelima berfungsi untuk menampilkan nilai PI dan sebagai tempat input nilai jari-jari lingkaran.
3. Barisan keenam berfungsi untuk menghitung nilai luas lingkaran dengan PI dan nilai jari-jari lingkaran yang telah diinput.
4. Barisan ketujuh berfungsi untuk menampilkan hasil perhitungan luas lingkaran.
