## 2.1 Dasar Teori

1. **Struktur Perulangan for dalam Python**
   Perulangan for digunakan untuk melakukan iterasi terhadap suatu urutan (seperti list, tuple, string, atau rentang angka menggunakan range()). Struktur ini memungkinkan perintah dieksekusi berulang kali sesuai dengan jumlah iterasi yang ditentukan.

2. **Perulangan Bersarang (*Nested Loop*)**
   *Nested loop* adalah struktur perulangan di mana sebuah blok perulangan berada di dalam blok perulangan lainnya. Perulangan terluar (*outer loop*) mengontrol baris atau tingkatan utama, sedangkan perulangan terdalam (*inner loop*) mengontrol elemen-elemen atau kolom pada setiap baris tersebut.

3. **Pengontrolan Parameter Output print()**
   Dalam pencetakan pola, parameter end="" pada fungsi print() digunakan untuk mencegah perpindahan baris otomatis, sehingga karakter berikutnya dicetak pada baris yang sama. Perintah print() tanpa argumen digunakan untuk melakukan perpindahan baris baru (*newline*).

## 2.2 Kode Program Latihan Praktikum

### 2.2.1 Program perulangan_for1_2611532008

```python
ulang_2008 = int(input("Masukkan jumlah perulangan: "))

for i_2008 in range(ulang_2008):
    print(f"Perulangan ke-{i_2008}")
```

**Penjelasan :**
1. Mengambil input dari pengguna sebagai batas akumulasi perulangan (**ulang_2008**).
2. Melakukan perulangan menggunakan **range(ulang_2008)** yang dimulai dari indeks 0 hingga **ulang_2008 - 1**.
3. Mencetak kalimat status perulangan beserta indeks variabel **i_2008** pada setiap iterasi

### 2.2.2 Program perulangan_for2_2611532008.py

```python
ulang_2008 = int(input("Masukkan jumlah perulangan: "))
print("Perulangan ke-0 sampai ke-", ulang_2008-1)
for i_2008 in range(ulang_2008):
    print(i_2008, end="")
print()
print("Perulangan ke-1 sampai ke-", ulang_2008)
for i_2008 in range(1, ulang_2008+1):
    print(i_2008, end=" ")
```

**Penjelasan :**
1. Mengambil masukan batas perulangan **ulang_2008** dari keyboard.
2. Loop pertama menampilkan nilai iterasi dari basis 0 (**0  hingga ulang_2008 - 1**) tanpa jarak antar karakter (**end=""**).
3. Loop kedua menampilkan nilai iterasi dari basis 1 (**1 hingga ulang_2008**) yang dipisahkan spasi (**end=" "**).

### 2.2.3 Program perulangan_for3_2611532008.py

```python
ulang_2008 = int(input("Masukkan jumlah perulangan: "))

jumlah_2008 = 0
for i_2008 in range(1, ulang_2008+1):
    print(i_2008, end=" ")
    jumlah_2008 = jumlah_2008 + i_2008

    if i_2008 < ulang_2008:
        print("+", end=" ")
    else:
        print(" = ", jumlah_2008, end=" ")
print()
print("Jumlah =", jumlah_2008)
```

**Penjelasan :**
1. Menginisialisasi akumulator total jumlah_2008 = 0.
2. Menjumlahkan setiap bilangan i_2008 secara bertahap dari 1 hingga ulang_2008.
3. Menggunakan pengondisian if-else untuk menyusun format deret penjumlahan matematis (menampilkan tanda + di antara angka dan tanda = di akhir deret).

### 2.2.4 Program jumlah_genap_2611532008.py

```python
ulang_2008 = int(input("Masukkan nilai batas: "))

jumlah_2008 = 0
for i in range(1, ulang_2008+1):
    if i % 2 == 0:
        print(i, end=" ")
        jumlah_2008 = jumlah_2008 + i

        if i < ulang_2008:
            print("+", end=" ")
        else:
            print(" = ", jumlah_2008, end=" ")
print()
print("Jumlah =", jumlah_2008)
```

**Penjelasan :**
1. Memeriksa setiap nilai i dari 1 sampai ulang_2008 menggunakan operator modulus i % 2 == 0 untuk menyaring bilangan genap.
2. Hanya menambahkan dan mencetak angka yang bernilai genap ke dalam variabel akumulator jumlah_2008.

### 2.2.5 Program nested_for1_2611532008.py

```python
batas_2008 = int(input("Masukkan batas: "))
for line_2008 in range(1, batas_2008+1):
    for j_2008 in range(1, (-1 * line_2008 + batas_2008 + 1)):
        print(" . ", end="")
    print(line_2008)
```

**Penjelasan :**
1. Outer loop (line_2008) mengontrol jumlah baris dari 1 sampai batas_2008.
2. Inner loop (j_2008) mencetak karakter titik " . " yang jumlahnya berkurang secara berkala sesuai rumus (-1 * line_2008 + batas_2008 + 1).
3. Menampilkan nomor baris di akhir setiap deret titik

### 2.2.6 Program nested_for2_2611532008.py

```python
batas_2008 = int(input("Masukkan nilai batas: "))
for i_2008 in range(1, batas_2008 + 1):
    for j_2008 in range(1, batas_2008 + 1):
        print("*", end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**
1. Menggunakan dua perulangan bersarang untuk mencetak matriks simetris.
2. Inner loop mencetak tanda * sejumlah batas_2008, dan outer loop mengulangnya sebanyak batas_2008 baris

### 2.2.7 Program nested_for3_2611532008.py

```python
batas_2008 = int(input("Masukkan nilai batas: "))
for i_2008 in range(batas_2008+1):
    for j_2008 in range(batas_2008+1):
        print(i_2008+j_2008, end="  ")
    print()  # pindah ke baris berikutnya
```

**Penjelasan :**
1. Melakukan perulangan indeks dari 0 hingga batas_2008 untuk outer (i_2008) dan inner loop (j_2008).
2. Menampilkan hasil penjumlahan indeks baris dan kolom (i_2008 + j_2008) sehingga membentuk pola tabel penjumlahan.

### 2.2.8 Program nested_for4_2611532008.py

```python
tinggi_2008 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2008 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2008 = tinggi_2008
    c_2008 = a_2008
    lebar_2008 = (2 * tinggi_2008) - 2

    for i_2008 in range(1, tinggi_2008 + 1):
        b_2008 = c_2008 + 1

        for j_2008 in range(1, lebar_2008 +1):

            # Baris atas dan bawah
            if i_2008 == 1 or i_2008 == tinggi_2008:
                if j_2008 == 1 or j_2008 == lebar_2008:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j_2008 == 1 or j_2008 == lebar_2008:
                    print("|", end="")
                else:
                    if j_2008 == c_2008:
                        print("<", end="")
                    elif j_2008 == b_2008:
                        print(">", end="")
                    elif j_2008 == (lebar_2008 - c_2008):
                        print("<", end="")
                    elif j_2008 == (lebar_2008 - c_2008 +1):
                        print(">", end="")
                    elif j_2008 > b_2008 and j_2008 < (lebar_2008 - c_2008):
                        print(",", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika pembentukan pola simetris
        a_2008 -= 2

        if a_2008 <= 0:
            c_2008 = (-a_2008) + 2
        else: 
            c_2008 = a_2008 
```

**Penjelasan :**
1. Program di atas menerima masukan tinggi berukuran genap. 
2. Program mengombinasikan *nested loop* dengan pengecekan posisi koordinat "(i_2008, j_2008)" untuk membentuk bingkai karakter bertepi "#", "=", dan "|". 
3. memuat bentuk simetris internal seperti "<", ">", dan "," di bagian dalamnya.

### 2.2.9 Program Latihan5_2611532008.py

```python
tinggi_2008 = int(input("Masukkan tinggi segitiga: "))
for i_2008 in range(1, tinggi_2008 + 1):
    print(" " * (tinggi_2008 - i_2008), end="")
    
    for j_2008 in range(i_2008):
        print("*", end=" ")
    print() # pindah ke baris berikutnya
```

**Penjelasan :**
1. Mengatur jarak spasi di sebelah kiri segitiga dengan perkalian string " " * (tinggi_2008 - i_2008) agar posisi bintang rata tengah.
2. Inner loop mencetak karakter "* " sebanyak nilai baris i_2008 untuk membentuk segitiga.

## Tugas Praktikum: PROGRAM JAM PASIR KRISTAL PALINDROMIK

**Soal :**
  Buatlah program Python yang mencetak Pola Jam Pasir Kristal Palindromik Berbingkai secara dinamis menggunakan perulangan for bersarang murni dengan nama berkas tugas5_nim.py (ganti nim dengan nomor induk mahasiswa masing-masing).

**Kode Program :**

Program `framed_crystal_hourglass_2611532008.py` (Murni tanpa fungsi `def`):

```python
# Program Jam Pasir Kristal Palindromik
print ("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK ===")
n_2008 = int(input("Masukkan tinggi jam pasir (N): "))

# 1. Bingkai Pembatas Horizontal Atas
lebar_tengah_2008 = 4 * n_2008 + 5
print("#" + "=" * lebar_tengah_2008 + "#")

# 2. Fase 1: Jam Pasir Atas (N s.d. 1)
for baris_2008 in range(n_2008, 0, -1):
    print("| " + " " * (2 * (n_2008 - baris_2008)), end="")
    
    for k_2008 in range(baris_2008, 0, -1):
        print(f"{k_2008} ", end="")
        
    print("<*>", end="")
    
    for k_2008 in range(1, baris_2008 + 1):
        print(f" {k_2008}", end="")
        
    print(" " * (2 * (n_2008 - baris_2008)) + " |")

# 3. Fase 2: Poros Titik Pusat Jam Pasir (Singularity)
spasi_2008 = 2 * n_2008 + 1
print("|" + " " * spasi_2008 + "<*>" + " " * spasi_2008 + "|")

# 4. Fase 3: Jam Pasir Bawah (1 s.d. N)
for baris_2008 in range(1, n_2008 + 1):
    print("| " + " " * (2 * (n_2008 - baris_2008)), end="")
    
    for k_2008 in range(baris_2008, 0, -1):
        print(f"{k_2008} ", end="")
        
    print("<*>", end="")
    
    for k_2008 in range(1, baris_2008 + 1):
        print(f" {k_2008}", end="")
        
    print(" " * (2 * (n_2008 - baris_2008)) + " |")

# 5. Bingkai Pembatas Horizontal Bawah
print("#" + "=" * lebar_tengah_2008 + "#")
```

**Penjelasan Program Tugas:**
1. **Penerimaan Masukan:** Program menerima nilai n_2008 sebagai skala ukuran jam pasir.
2. **Bingkai Atas dan Bawah:** Panjang garis tengah sama dengan 4 * n_2008 + 5 yang diapit karakter sudut #.
3. **Fase 1 (Jam Pasir Atas):** Menggunakan perulangan for baris_2008 in range(n_2008, 0, -1). Pada setiap baris, program mencetak spasi penyeimbang "2 * (n - baris)", deret mundur k_2008 dari baris_2008 ke 1, simbol poros <*>, dan deret maju 1 ke baris_2008.
4. **Fase 2 (Titik Pusat):** Mencetak satu baris khusus berisi poros kristal <*> di tengah-tengah dengan spasi penyeimbang 2 * n + 1 di kiri dan kanan.
5. **Fase 3 (Jam Pasir Bawah):** Menggunakan perulangan for baris_2008 in range(1, n_2008 + 1) dengan logika pencetakan palindromik yang sama untuk mengekspansi pola kembali ke ukuran n.
